# Notebook 03 — YOLO11n (Nano) Training
**Dataset:** A (Drone/UAV) | **Paradigm:** One-Stage CNN | **Target:** Edge deployment

## Config
- Image: 512×512 | Batch: 8 | Epochs: 100 | Optimizer: AdamW | FP16: ✅
- Early stopping: patience=15 | Augmentation: Mosaic ON

```python
from ultralytics import YOLO
from pathlib import Path
ROOT = Path('..')
DATA_YAML = ROOT / 'data' / 'dataset-a' / 'data.yaml'
RESULTS_DIR = ROOT / 'results' / 'experiment-a' / 'yolo11n'
```

```python
# data.yaml — YOLO11 requires this file
yaml_content = f'''
path: {(ROOT / 'data' / 'dataset-a').resolve()}
train: images/train/images
val: images/val/images
nc: 2
names: ['fire', 'smoke']
'''
DATA_YAML.write_text(yaml_content)
print(DATA_YAML.read_text())
```

```python
model = YOLO('yolo11n.pt')  # loads pretrained COCO weights

results = model.train(
    data=str(DATA_YAML),
    epochs=100,
    batch=8,
    imgsz=512,
    optimizer='AdamW',
    lr0=0.001,
    half=True,           # FP16
    mosaic=1.0,          # mosaic augmentation
    patience=15,         # early stopping
    project=str(RESULTS_DIR),
    name='train',
    exist_ok=True,
    device=0             # GPU
)
```

```python
# ── Evaluation ────────────────────────────────────────────────────────────
metrics = model.val(data=str(DATA_YAML), imgsz=512, half=True)
print(f"mAP@0.5:      {metrics.box.map50:.4f}")
print(f"Precision:    {metrics.box.mp:.4f}")
print(f"Recall:       {metrics.box.mr:.4f}")
```

```python
# ── FPS Benchmark ─────────────────────────────────────────────────────────
import time, torch
model_eval = YOLO(RESULTS_DIR / 'train' / 'weights' / 'best.pt')
dummy = torch.zeros(1, 3, 512, 512).cuda()
# warmup
for _ in range(5): model_eval.predict(source=dummy, verbose=False)
# benchmark
start = time.time()
N = 50
for _ in range(N): model_eval.predict(source=dummy, verbose=False)
fps = N / (time.time() - start)
print(f"FPS on T4: {fps:.1f}")
```
